```python
import math
from types import NotImplementedType
```

We import the math module to use math.isclose()
when comparing temperatures. This allows small
floating-point differences.

```python

class Temperature:
    """
     A class for representing temperatures in Celsius, Fahrenheit and Kelvin.
    """    
    def __init__(self, value: float = 0.0, unit: str =  'C')-> None:
        """
            Initialize a temperature object.
        
            Args:
                value: The numerical temperature value.
                unit: The temperature unit (C, F or K).
                          Defaults to Celsius.
        
            Raises:
                ValueError: If the unit is invalid or the
                            temperature is below absolute zero.
        """
        # Store the temperature value and normalize the unit.
        self._value = value
        self._unit = unit.upper()

        # Check that the temperature unit is valid.
        valid_units = ['C', 'F', 'K']
        if self._unit not in valid_units:
            raise ValueError(f"Invalid unit '{self._unit}'. Valid units are: {', '.join(valid_units)}.")
        
        # Check that the temperature is not below absolute zero.
        if self._unit == 'C' and self._value < -273.15:
            raise ValueError("Temperature cannot be below absolute zero (-273.15°C).")
        elif self._unit == 'F' and self._value < -459.67:
            raise ValueError("Temperature cannot be below absolute zero (-459.67°F).")
        elif self._unit == 'K' and self._value < 0:
            raise ValueError("Temperature cannot be below absolute zero (0 K).")

    @property
    def value(self)-> float:
        """Return the numerical temperature value."""
        return self._value
    @property
    def unit(self)-> str:
        """Return the temperature unit."""
        return self._unit

    def to_celsius(self)-> float:
        """Convert the temperature to Celsius."""
        if self._unit == 'C':
            return self._value
        elif self._unit == 'F':
            return (self._value - 32) * 5.0 / 9.0
        elif self._unit == 'K':
            return self._value - 273.15
       

    def to_fahrenheit(self)-> float:
        """Convert the temperature to Fahrenheit."""
        if self._unit == 'C':
            return (self._value * 9.0 / 5.0) + 32
        elif self._unit == 'F':
            return self._value
        elif self._unit == 'K':
            return (self._value - 273.15) * 9.0 / 5.0 + 32
   

    def to_kelvin(self)-> float:
        """Convert the temperature to Kelvin."""
        if self._unit == 'C':
            return self._value + 273.15
        elif self._unit == 'F':
            return (self._value - 32) * 5.0 / 9.0 + 273.15
        elif self._unit == 'K':
            return self._value

    def convert_to(self, new_unit: str) -> 'Temperature':
        """
        Convert the temperature to a new unit.

        Args:
            new_unit: The unit to convert to (C, F or K).

        Returns:
            A new Temperature object in the requested unit.

        Raises:
            ValueError: If the requested unit is invalid.
        """

        new_unit = new_unit.upper()
        # Validate the requested unit.
        if new_unit not in ['C', 'F', 'K']:
            raise ValueError(f"Invalid unit '{new_unit}'. Valid units are: C, F, K.")

        # Return a new Temperature object in the requested unit.
        if new_unit == 'C':
            return Temperature(self.to_celsius(), 'C')
        elif new_unit == 'F':
            return Temperature(self.to_fahrenheit(), 'F')
        elif new_unit == 'K':
            return Temperature(self.to_kelvin(), 'K')
        
    def __eq__(self, other: object)-> bool:
        """Check if two temperatures are equal."""
        if not isinstance(other, Temperature):
            return NotImplemented
        return math.isclose(
                self.to_celsius(),
                other.to_celsius(),
                rel_tol=1e-9,
                abs_tol=1e-9
            )
    
    def __ne__(self, other: object)-> bool:
        """Check if two temperatures are not equal."""
        if not isinstance(other, Temperature):
            return NotImplemented
        return not math.isclose(
            self.to_celsius(),
            other.to_celsius(),
            rel_tol=1e-9,
            abs_tol=1e-9
        )
    
    def __lt__(self, other: object)-> bool:
        """Check if the temperature is less than another."""
        if not isinstance(other, Temperature):
            return NotImplemented
        return self.to_celsius() < other.to_celsius()

    def __le__(self, other: object)-> bool:
        """Check if the temperature is less than or equal to another."""
        if not isinstance(other, Temperature):
            return NotImplemented
        return self.to_celsius() <= other.to_celsius()

    def __gt__(self, other: object)-> bool:
        """Check if the temperature is greater than another."""
        if not isinstance(other, Temperature):
            return NotImplemented
        return self.to_celsius() > other.to_celsius()

    def __ge__(self, other: object)-> bool:
        """Check if the temperature is greater than or equal to another."""
        if not isinstance(other, Temperature):
            return NotImplemented
        return self.to_celsius() >= other.to_celsius()
    


    def __str__(self)-> str:
        """Return a readable string with two decimal places."""
        return f"{self._value:.2f} °{self._unit}"
    
    def __repr__(self)-> str:
        """Return a string representation of the Temperature object."""
        return f"Temperature(value={self._value}, unit='{self._unit}')"

```

The constructor initializes a temperature object.
It uses 0.0 Celsius as the default value and C as
the default unit.

The constructor also checks that the unit is valid
and that the temperature is not below absolute zero.

# Task 2: Fractions

In this task, we implement a Fraction class to represent fractions using a numerator and a denominator.

The class supports comparisons between fractions, arithmetic operations, and string representations.

The numerator and denominator must be integers, and the denominator must be strictly positive.

### Implementation

The Fraction class is implemented in a separate Python file, `fraction.py`, and imported into this notebook.
The class uses special methods to support comparisons, arithmetic operations, and string representations.
The implementation includes type hints, docstrings, and input validation to ensure that fractions are created correctly.
The following sections demonstrate the functionality of the class using different examples.

The denominator must be positive to ensure a consistent representation of fractions. Negative fractions are represented with the negative sign in the numerator.
This also allows comparisons using cross multiplication without reversing the comparison operators.

```python
from fraction import Fraction

# Create three fraction objects.
fraction1 = Fraction(1, 2)
fraction2 = Fraction(2, 4)
fraction3 = Fraction(3, 4)

# Display the string representations.
print("Fraction:", fraction1)
print("Representation:", repr(fraction1))
```

## Comparing fractions

The Fraction class implements all six comparison methods.
Fractions are compared using cross multiplication, allowing fractions with different numerators and denominators to be compared without converting them to floating-point numbers.

For example, 1/2 and 2/4 are equal even though their numerators and denominators are different.

```python
# Compare fractions using the six comparison operators.
print("Equal:", fraction1 == fraction2)
print("Not equal:", fraction1 != fraction3)
print("Less than:", fraction1 < fraction3)
print("Less than or equal:", fraction1 <= fraction2)
print("Greater than:", fraction3 > fraction1)
print("Greater than or equal:", fraction3 >= fraction2)
```

## Arithmetic operations

The Fraction class supports four arithmetic operations: addition, subtraction, multiplication, and division.
Each operation returns a new Fraction object without changing the original fractions.

The fractions are not automatically simplified. We keep the numerator and denominator produced by each arithmetic operation, which makes the calculations easier to follow.

Division by a fraction with a numerator of zero raises a ZeroDivisionError.

```python
# Perform arithmetic operations using two fractions.
print("Addition:", fraction1 + fraction3)
print("Subtraction:", fraction1 - fraction3)
print("Multiplication:", fraction1 * fraction3)
print("Division:", fraction1 / fraction3)
```

## Error handling and validation

The Fraction class validates the numerator and denominator when a new fraction is created.

Both values must be integers, and the denominator must be strictly positive.

Invalid values raise TypeError or ValueError.

The class also prevents division by zero by raising a ZeroDivisionError.

```python
# Check valid fractions with zero and negative numerators.
print("Zero numerator:", Fraction(0, 3))
print("Negative quotient:", Fraction(1, 2) / Fraction(-1, 3))

# Check an invalid denominator.
try:
    Fraction(1, 0)
except ValueError as error:
    print("Invalid denominator:", error)

# Check a negative denominator.
try:
    Fraction(1, -2)
except ValueError as error:
    print("Negative denominator:", error)

# Check an invalid numerator.
try:
    Fraction(1.5, 2)
except TypeError as error:
    print("Invalid numerator:", error)

# Check division by zero.
try:
    Fraction(1, 2) / Fraction(0, 3)
except ZeroDivisionError as error:
    print("Division by zero:", error)
```
